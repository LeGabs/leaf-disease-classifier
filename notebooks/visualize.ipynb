```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("../data/PlantVillage/Pepper__bell___Bacterial_spot/0a0dbf1f-1131-496f-b337-169ec6693e6f___NREC_B.Spot 9241.JPG")
```

### Histogram Pepper Bacterial Spot

```python
img_hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

hist_h = cv2.calcHist([img_hsv], [0], None, [179], [0, 179])



plt.plot(hist_h, label='Hue channel')
plt.xlabel('Pixel Value')
plt.ylabel('Frequency')
plt.title('Histogram of HSV Channels')
plt.legend()
plt.show()

```
